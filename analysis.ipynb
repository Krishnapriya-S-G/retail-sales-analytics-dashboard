```python
df_cancelled = df[df['Invoice'].astype(str).str.startswith('C')]

df_clean = df[~df['Invoice'].astype(str).str.startswith('C')]

print(f"Cancelled orders removed: {df_cancelled.shape[0]}")
print(f"Remaining rows: {df_clean.shape[0]}")
```

```python
missing_pct = (df_clean['Customer ID'].isnull().sum() / df_clean.shape[0]) * 100
print(f"Missing Customer ID percentage: {missing_pct:.2f}%")
```

```python
df_clean['Customer ID'] = df_clean['Customer ID'].fillna(0)
```

```python
df_clean = df_clean.dropna(subset=['Description'])
print(f"Rows after dropping missing descriptions: {df_clean.shape[0]}")
```

```python
df_clean = df_clean.drop_duplicates()
print(f"Rows after removing duplicates: {df_clean.shape[0]}")
```

```python
df_clean = df_clean[(df_clean['Quantity'] > 0) & (df_clean['Price'] > 0)]
print(f"Final cleaned rows: {df_clean.shape[0]}")
```

```python
df_clean.isnull().sum()
```

```python
df_clean['TotalSales'] = df_clean['Quantity'] * df_clean['Price']
df_clean[['Quantity', 'Price', 'TotalSales']].head()
```

```python
df_clean['InvoiceDate'] = pd.to_datetime(df_clean['InvoiceDate'])

df_clean['Year'] = df_clean['InvoiceDate'].dt.year
df_clean['Month'] = df_clean['InvoiceDate'].dt.month
df_clean['Day'] = df_clean['InvoiceDate'].dt.day
df_clean['Weekday'] = df_clean['InvoiceDate'].dt.day_name()
df_clean['Hour'] = df_clean['InvoiceDate'].dt.hour

df_clean[['InvoiceDate', 'Year', 'Month', 'Day', 'Weekday', 'Hour']].head()
```

```python
df_clean['TotalSales'] = df_clean['Quantity'] * df_clean['Price']
df_clean[['Quantity', 'Price', 'TotalSales']].head()
```

```python
df_clean.head()
```

```python
df_clean.to_csv('cleaned_retail_data.csv', index=False)
print("File saved successfully!")
print(f"Final shape: {df_clean.shape}")
```

```python
import sqlite3

conn = sqlite3.connect('retail_data.db')

df_clean.to_sql('sales', conn, if_exists='replace', index=False)

print("Database created successfully!")
```

```python
query = "SELECT * FROM sales LIMIT 5"
result = pd.read_sql(query, conn)
result
```

```python
query1_clean = """
SELECT Description, 
       SUM(TotalSales) as Revenue,
       SUM(Quantity) as TotalQuantitySold
FROM sales
WHERE Description NOT IN ('Manual', 'DOTCOM POSTAGE', 'POSTAGE', 'CARRIAGE', 'Discount')
GROUP BY Description
ORDER BY Revenue DESC
LIMIT 10
"""
result1_clean = pd.read_sql(query1_clean, conn)
result1_clean
```

```python
query2 = """
SELECT Year, Month, 
       SUM(TotalSales) as MonthlyRevenue
FROM sales
GROUP BY Year, Month
ORDER BY Year, Month
"""
result2 = pd.read_sql(query2, conn)
result2
```

```python
query3 = """
SELECT Country, 
       SUM(TotalSales) as Revenue,
       COUNT(DISTINCT Invoice) as NumberOfOrders
FROM sales
GROUP BY Country
ORDER BY Revenue DESC
LIMIT 10
"""
result3 = pd.read_sql(query3, conn)
result3
```

```python
query4 = """
SELECT "Customer ID",
       COUNT(DISTINCT Invoice) as Frequency,
       SUM(TotalSales) as Monetary,
       MAX(InvoiceDate) as LastPurchaseDate
FROM sales
WHERE "Customer ID" != 0
GROUP BY "Customer ID"
ORDER BY Monetary DESC
LIMIT 10
"""
result4 = pd.read_sql(query4, conn)
result4
```
